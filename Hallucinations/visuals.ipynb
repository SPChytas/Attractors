```python
original = pd.read_json('/data/takis/Hallucinations/CHAIR/llava/layer_17_strength_0.000_tokens_512_keep_300.jsonl', lines=True)
modified = pd.read_json('/data/takis/Hallucinations/CHAIR/llava/layer_20_strength_11.000_tokens_512_keep_300.jsonl', lines=True)

modified.head()
```

```python
or_len = [len(row['caption']) for _, row in original.iterrows()]
mod_len = [len(row['caption']) for _, row in modified.iterrows()]
```

```python
rcParams['figure.figsize'] = (1.5, 3)


b = plt.boxplot(or_len, widths=[0.4], positions=[0], vert=True, medianprops={'color': 'black'}, flierprops={'color': 'black','mfc': '#a6a2a2'}, patch_artist=True)
for patch in b['boxes']:
    patch.set_facecolor('#a6a2a2')

b = plt.boxplot(mod_len, widths=[0.4], positions=[0.5], vert=True, medianprops={'color': 'black'}, flierprops={'color': 'black','mfc': '#2b9348'}, patch_artist=True)
for patch in b['boxes']:
    patch.set_facecolor('#2b9348')

plt.xlim((-0.3, 0.8))
plt.ylabel('Length')
plt.xticks([])
plt.show()
```
